```python
team_stats = {}

team1_win_rates = []
team2_win_rates = []

for _, row in ml_data.iterrows():

    team1 = row['team1']
    team2 = row['team2']

    # Get stats BEFORE this match
    team1_stats = team_stats.get(
        team1,
        {'matches': 0, 'wins': 0}
    )

    team2_stats = team_stats.get(
        team2,
        {'matches': 0, 'wins': 0}
    )

    # Calculate pre-match win rates
    team1_rate = (
        team1_stats['wins'] / team1_stats['matches']
        if team1_stats['matches'] > 0
        else 0.5
    )

    team2_rate = (
        team2_stats['wins'] / team2_stats['matches']
        if team2_stats['matches'] > 0
        else 0.5
    )

    team1_win_rates.append(team1_rate)
    team2_win_rates.append(team2_rate)

    # Update statistics AFTER the match
    team1_stats['matches'] += 1
    team2_stats['matches'] += 1

    if row['winner'] == team1:
        team1_stats['wins'] += 1
    elif row['winner'] == team2:
        team2_stats['wins'] += 1

    team_stats[team1] = team1_stats
    team_stats[team2] = team2_stats
```

```python
recent_form = {}

team1_recent = []
team2_recent = []

for _, row in ml_data.iterrows():

    team1 = row['team1']
    team2 = row['team2']

    team1_history = recent_form.get(team1, [])
    team2_history = recent_form.get(team2, [])

    # Calculate form BEFORE current match
    team1_form = (
        sum(team1_history[-5:]) / len(team1_history[-5:])
        if team1_history
        else 0.5
    )

    team2_form = (
        sum(team2_history[-5:]) / len(team2_history[-5:])
        if team2_history
        else 0.5
    )

    team1_recent.append(team1_form)
    team2_recent.append(team2_form)

    # Update AFTER current match
    team1_history.append(1 if row['winner'] == team1 else 0)
    team2_history.append(1 if row['winner'] == team2 else 0)

    recent_form[team1] = team1_history
    recent_form[team2] = team2_history
```

```python
venue_stats = {}

venue_chasing_rates = []

for _, row in ml_data.iterrows():

    venue = row['venue']

    stats = venue_stats.get(
        venue,
        {'matches': 0, 'chasing_wins': 0}
    )

    if stats['matches'] > 0:
        venue_rate = (
            stats['chasing_wins'] / stats['matches']
        )
    else:
        venue_rate = 0.5

    venue_chasing_rates.append(venue_rate)

    # Update AFTER the current match
    stats['matches'] += 1

    if row['win_by_wickets'] > 0:
        stats['chasing_wins'] += 1

    venue_stats[venue] = stats
```

```python
ml_data['venue_chasing_win_rate'] = venue_chasing_rates
```

```python
h2h_stats = {}
team1_h2h_rates = []

for _, row in ml_data.iterrows():

    team1 = row['team1']
    team2 = row['team2']

    pair = tuple(sorted([team1, team2]))

    stats = h2h_stats.get(
        pair,
        {'matches': 0, 'team1_wins': 0, 'team2_wins': 0}
    )

    # Determine which team's historical wins belong to
    if stats['matches'] == 0:
        h2h_rate = 0.5

    else:
        if team1 == pair[0]:
            h2h_rate = stats['team1_wins'] / stats['matches']
        else:
            h2h_rate = stats['team2_wins'] / stats['matches']

    team1_h2h_rates.append(h2h_rate)

    # Update AFTER current match
    stats['matches'] += 1

    if row['winner'] == team1:
        if team1 == pair[0]:
            stats['team1_wins'] += 1
        else:
            stats['team2_wins'] += 1

    elif row['winner'] == team2:
        if team2 == pair[0]:
            stats['team1_wins'] += 1
        else:
            stats['team2_wins'] += 1

    h2h_stats[pair] = stats
```

```python
ml_data['team1_h2h_win_rate'] = team1_h2h_rates
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    'team1',
    'team2',
    'venue'
]

numeric_features = [
    'toss_winner_team1',
    'toss_decision_field',
    'team1_historical_win_rate',
    'team2_historical_win_rate',
    'team1_recent_form',
    'team2_recent_form',
    'venue_chasing_win_rate',
    'team1_h2h_win_rate'
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_features
        ),
        (
            'numeric',
            'passthrough',
            numeric_features
        )
    ]
)
```

```python
X_train_encoded = preprocessor.fit_transform(X_train)

X_test_encoded = preprocessor.transform(X_test)
```

```python
import pandas as pd

def fix_ipl_date(x):
    if pd.isna(x):
        return pd.NaT

    # Dates that Excel already converted incorrectly
    if isinstance(x, pd.Timestamp):
        return pd.Timestamp(
            year=x.year,
            month=x.day,
            day=x.month
        )

    # Dates that remained as DD-MM-YYYY text
    return pd.to_datetime(x, dayfirst=True)


matches['date'] = matches['date'].apply(fix_ipl_date)
```

```python
ml_data = matches_clean.copy()

ml_data = ml_data[ml_data['winner'].notna()].copy()

ml_data['target'] = (
    ml_data['winner'] == ml_data['team1']
).astype(int)

ml_data['toss_winner_team1'] = (
    ml_data['toss_winner'] == ml_data['team1']
).astype(int)

ml_data['toss_decision_field'] = (
    ml_data['toss_decision'] == 'field'
).astype(int)
```

```python
team_stats = {}

team1_win_rates = []
team2_win_rates = []

for _, row in ml_data.iterrows():

    team1 = row['team1']
    team2 = row['team2']

    team1_stats = team_stats.get(
        team1, {'matches': 0, 'wins': 0}
    )

    team2_stats = team_stats.get(
        team2, {'matches': 0, 'wins': 0}
    )

    team1_rate = (
        team1_stats['wins'] / team1_stats['matches']
        if team1_stats['matches'] > 0 else 0.5
    )

    team2_rate = (
        team2_stats['wins'] / team2_stats['matches']
        if team2_stats['matches'] > 0 else 0.5
    )

    team1_win_rates.append(team1_rate)
    team2_win_rates.append(team2_rate)

    team1_stats['matches'] += 1
    team2_stats['matches'] += 1

    if row['winner'] == team1:
        team1_stats['wins'] += 1
    elif row['winner'] == team2:
        team2_stats['wins'] += 1

    team_stats[team1] = team1_stats
    team_stats[team2] = team2_stats

ml_data['team1_historical_win_rate'] = team1_win_rates
ml_data['team2_historical_win_rate'] = team2_win_rates
```

```python
recent_form = {}

team1_recent = []
team2_recent = []

for _, row in ml_data.iterrows():

    team1 = row['team1']
    team2 = row['team2']

    team1_history = recent_form.get(team1, [])
    team2_history = recent_form.get(team2, [])

    team1_recent.append(
        sum(team1_history[-5:]) / len(team1_history[-5:])
        if team1_history else 0.5
    )

    team2_recent.append(
        sum(team2_history[-5:]) / len(team2_history[-5:])
        if team2_history else 0.5
    )

    team1_history.append(1 if row['winner'] == team1 else 0)
    team2_history.append(1 if row['winner'] == team2 else 0)

    recent_form[team1] = team1_history
    recent_form[team2] = team2_history

ml_data['team1_recent_form'] = team1_recent
ml_data['team2_recent_form'] = team2_recent
```

```python
venue_stats = {}
venue_chasing_rates = []

for _, row in ml_data.iterrows():

    venue = row['venue']

    stats = venue_stats.get(
        venue,
        {'matches': 0, 'chasing_wins': 0}
    )

    venue_rate = (
        stats['chasing_wins'] / stats['matches']
        if stats['matches'] > 0 else 0.5
    )

    venue_chasing_rates.append(venue_rate)

    stats['matches'] += 1

    if row['win_by_wickets'] > 0:
        stats['chasing_wins'] += 1

    venue_stats[venue] = stats

ml_data['venue_chasing_win_rate'] = venue_chasing_rates
```

```python
h2h_stats = {}
team1_h2h_rates = []

for _, row in ml_data.iterrows():

    team1 = row['team1']
    team2 = row['team2']

    pair = tuple(sorted([team1, team2]))

    stats = h2h_stats.get(
        pair,
        {'matches': 0, 'team1_wins': 0, 'team2_wins': 0}
    )

    if stats['matches'] == 0:
        h2h_rate = 0.5
    else:
        if team1 == pair[0]:
            h2h_rate = stats['team1_wins'] / stats['matches']
        else:
            h2h_rate = stats['team2_wins'] / stats['matches']

    team1_h2h_rates.append(h2h_rate)

    stats['matches'] += 1

    if row['winner'] == team1:
        if team1 == pair[0]:
            stats['team1_wins'] += 1
        else:
            stats['team2_wins'] += 1

    elif row['winner'] == team2:
        if team2 == pair[0]:
            stats['team1_wins'] += 1
        else:
            stats['team2_wins'] += 1

    h2h_stats[pair] = stats

ml_data['team1_h2h_win_rate'] = team1_h2h_rates
```
